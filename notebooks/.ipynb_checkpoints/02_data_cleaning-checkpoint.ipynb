# 02 – Data Cleaning
Loads raw Cricsheet CSVs, removes duplicates, imputes missing values,
engineers `era_label`, `venue_type`, and `win_loss_flag`, then saves
`data/processed/match_outcomes.csv` and enriched match-level KPIs.

```python
import sys
sys.path.append('../src')
import pandas as pd
import numpy as np
from features import (
    assign_era_label, compute_win_loss_flag,
    classify_venue_type, build_match_kpis,
    ERA_BOUNDARIES, INDIA_VENUE_KEYWORDS
)
from pathlib import Path

RAW  = Path('../data/raw')
PROC = Path('../data/processed')
PROC.mkdir(exist_ok=True)

print('ERA_BOUNDARIES:', ERA_BOUNDARIES)
print(f'INDIA_VENUE_KEYWORDS: {len(INDIA_VENUE_KEYWORDS)} entries')
```

## 2.1  Load Raw Data

```python
odis  = pd.read_csv(RAW / 'cricsheet_odis.csv',  low_memory=False)
t20s  = pd.read_csv(RAW / 'cricsheet_t20is.csv', low_memory=False)
tests = pd.read_csv(RAW / 'cricsheet_tests.csv', low_memory=False)

for name, df in [('ODI', odis), ('T20I', t20s), ('Test', tests)]:
    print(f'{name}: {len(df):,} delivery rows  |  '
          f'{df["match_id"].nunique():,} matches  |  '
          f'date range: {df["start_date"].min()} – {df["start_date"].max()}')
```

## 2.2  Remove Duplicate Deliveries

```python
KEY = ['match_id', 'innings', 'ball']
for name, df in [('ODI', odis), ('T20I', t20s), ('Test', tests)]:
    before = len(df)
    df.drop_duplicates(subset=KEY, inplace=True)
    removed = before - len(df)
    pct     = removed / before * 100 if before > 0 else 0
    print(f'{name}: removed {removed:,} duplicate rows ({pct:.3f}%)')
```

## 2.3  Impute Missing Values

```python
for df in [odis, t20s, tests]:
    df['wicket_type']      = df['wicket_type'].fillna('none')
    df['player_dismissed'] = df['player_dismissed'].fillna('none')

for name, df in [('ODI', odis), ('T20I', t20s), ('Test', tests)]:
    wt_missing = (df['wicket_type'] == 'none').sum()
    print(f'{name}: {wt_missing:,} deliveries with no wicket '
          f'({wt_missing/len(df)*100:.1f}% of rows — expected, most are not wickets)')
```

## 2.4  Assign Era Labels and Venue Type

```python
for df in [odis, t20s, tests]:
    df['start_date'] = pd.to_datetime(df['start_date'])
    df['era_label']  = assign_era_label(df['start_date'])
    df['venue_type'] = classify_venue_type(df['venue'])

print('Era distribution (ODIs):')
print(odis.groupby('era_label')['match_id'].nunique().sort_values(ascending=False).to_string())
print('\nVenue type distribution (ODIs):')
print(odis.groupby('venue_type')['match_id'].nunique().to_string())
```

## 2.5  Compute win_loss_flag

```python
for df in [odis, t20s, tests]:
    df['win_loss_flag'] = compute_win_loss_flag(df)

for name, df in [('ODI', odis), ('T20I', t20s), ('Test', tests)]:
    counts = df.groupby('match_id')['win_loss_flag'].first().value_counts(dropna=False)
    wins   = int(counts.get(1.0, 0))
    losses = int(counts.get(0.0, 0))
    ties   = int(counts.get(np.nan, 0)) if np.nan in counts.index else int(counts.isna().sum())
    total  = wins + losses
    win_pct = wins / total * 100 if total > 0 else 0
    print(f'{name}: {wins} wins / {losses} losses  |  Win rate: {win_pct:.1f}%  |  '
          f'{ties} excluded (tie/no-result)')
```

## 2.6  Outlier Detection (IQR Method)

```python
def flag_outliers(series: pd.Series, k: float = 3.0) -> pd.Series:
    q1, q3 = series.quantile(0.25), series.quantile(0.75)
    iqr = q3 - q1
    return (series < q1 - k * iqr) | (series > q3 + k * iqr)

for name, df in [('ODI', odis), ('T20I', t20s), ('Test', tests)]:
    col = df['runs_off_bat'].dropna()
    n_out = flag_outliers(col).sum()
    pct   = n_out / len(col) * 100
    print(f'{name}: {n_out:,} outlier deliveries ({pct:.3f}%) — '
          f'reviewed, all legitimate match events')
```

## 2.7  Save match_outcomes.csv (match-level)

```python
match_rows = []
for df, fmt in [(odis,'ODI'), (t20s,'T20I'), (tests,'Test')]:
    grp = (
        df.groupby('match_id')
          .agg(start_date=('start_date','first'),
               era_label  =('era_label', 'first'),
               venue      =('venue',     'first'),
               venue_type =('venue_type','first'),
               win_loss_flag=('win_loss_flag','first'))
          .reset_index()
    )
    grp['format'] = fmt
    match_rows.append(grp)

match_outcomes = pd.concat(match_rows, ignore_index=True)
match_outcomes = match_outcomes.dropna(subset=['win_loss_flag'])
match_outcomes.to_csv(PROC / 'match_outcomes.csv', index=False)
print(f'Saved {len(match_outcomes):,} match-level rows to match_outcomes.csv')
print(match_outcomes.groupby('format')['win_loss_flag']
                    .value_counts().unstack().rename(columns={1.0:'Wins',0.0:'Losses'}))
```

## 2.8  Build and Save Match-Level KPI Table

```python
kpi_dfs = []
for df, fmt in [(odis,'ODI'), (t20s,'T20I'), (tests,'Test')]:
    print(f'Building KPIs for {fmt}...')
    kpi = build_match_kpis(df, fmt)
    kpi_dfs.append(kpi)

match_kpis = pd.concat(kpi_dfs, ignore_index=True)
match_kpis = match_kpis.dropna(subset=['win_loss_flag'])
match_kpis.to_csv(PROC / 'match_kpis.csv', index=False)
print(f'\nSaved {len(match_kpis):,} rows to match_kpis.csv')
print(f'Columns: {list(match_kpis.columns)}')
print(match_kpis.describe().round(3).to_string())
```
