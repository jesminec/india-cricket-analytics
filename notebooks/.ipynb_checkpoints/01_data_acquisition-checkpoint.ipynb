## 1.2  Verify Raw Files

```python
import os
files = sorted(OUTPUT_DIR.glob('*.csv'))
print(f'Found {len(files)} CSV files in {OUTPUT_DIR}\n')
for f in files:
    size_kb = os.path.getsize(f) / 1024
    n_rows  = sum(1 for _ in open(f)) - 1  # subtract header
    print(f'{f.name:50s}  {size_kb:9.1f} KB   {n_rows:>9,} rows')
```

## Figure 1 – Architecture / End-to-End Workflow Diagram

Generates `fig1_architecture.png` used in the v2 final report.

```python
import matplotlib
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np
from pathlib import Path

FIGS = Path('../reports/figures')
FIGS.mkdir(exist_ok=True)

# ── colour palette ──────────────────────────────────────────────────────────
BLUE   = '#1a5276'
GREEN  = '#1e8449'
ORANGE = '#d35400'
RED    = '#c0392b'
TEAL   = '#0e6655'
PURPLE = '#6c3483'
GRAY   = '#566573'

fig, ax = plt.subplots(figsize=(10, 6))
ax.set_xlim(0, 10); ax.set_ylim(0, 7); ax.axis('off')
fig.patch.set_facecolor('white')

def box(ax, x, y, w, h, text, color, fontsize=9, text_color='white'):
    rect = mpatches.FancyBboxPatch((x, y), w, h, boxstyle="round,pad=0.1",
                                    facecolor=color, edgecolor='white', linewidth=1.5)
    ax.add_patch(rect)
    ax.text(x+w/2, y+h/2, text, ha='center', va='center',
            fontsize=fontsize, color=text_color, fontweight='bold',
            multialignment='center')

def arrow(ax, x1, y1, x2, y2):
    ax.annotate('', xy=(x2,y2), xytext=(x1,y1),
                arrowprops=dict(arrowstyle='->', color='#2c3e50', lw=1.5))

ax.text(5, 6.6, 'End-to-End Workflow of the Proposed System',
        ha='center', va='center', fontsize=12, fontweight='bold', color='#2c3e50')

# Row 1 – data sources
box(ax,0.2,5.6,2.2,0.7,'Cricsheet\nCSV2 Data',      BLUE, 8)
box(ax,2.8,5.6,2.2,0.7,'ESPNcricinfo\nStatsguru',    BLUE, 8)
box(ax,5.4,5.6,2.2,0.7,'india_cricket_\ndownloader_v4.py', GRAY, 8)
box(ax,7.8,5.6,1.9,0.7,'GitHub\nRepository',         GRAY, 8)

# Row 2 – processing notebooks
box(ax,1.0,4.2,2.5,0.75,'Notebook 01\nData Ingestion',   TEAL, 8)
box(ax,4.0,4.2,2.5,0.75,'Notebook 02\nData Cleaning',    TEAL, 8)
box(ax,7.0,4.2,2.5,0.75,'Notebooks 03-07\nEDA & Features',TEAL, 8)

# Row 3 – models
box(ax,0.2,2.8,1.7,0.75,'LR (L2)\nBaseline',  GREEN,  8)
box(ax,2.1,2.8,1.7,0.75,'Random\nForest',      GREEN,  8)
box(ax,4.0,2.8,1.7,0.75,'XGBoost\nDART',       GREEN,  8)
box(ax,5.9,2.8,1.7,0.75,'ARIMA\nRQ2',          PURPLE, 8)
box(ax,7.8,2.8,1.9,0.75,'LMM/PELT\nRQ5',       PURPLE, 8)

# Row 4 – ensemble + eval
box(ax,1.5,1.5,3.0,0.75,'Stacking Meta-Classifier\nAUC-ROC = 0.922', ORANGE, 9)
box(ax,5.5,1.5,3.5,0.75,'Evaluation: Accuracy, F1,\nAUC-ROC, OR, Cohen\'s d', GRAY, 8)

# Row 5 – output
box(ax,3.0,0.3,4.0,0.8,'Plotly Dash Dashboards\nBCCI Coaching & Selection Staff', RED, 9)

# Arrows
for xs,ys,xt,yt in [
    (1.3,5.6,2.25,4.95),(3.9,5.6,5.25,4.95),(6.5,5.6,7.25,4.95),
    (2.25,4.2,2.25,3.55),(5.25,4.2,5.25,3.55),(8.25,4.2,8.25,3.55),
    (1.05,2.8,3.0,2.25),(2.95,2.8,3.0,2.25),(4.85,2.8,3.0,2.25),
    (6.75,2.8,7.25,2.25),(8.75,2.8,7.25,2.25),
    (3.0,1.5,5.0,0.95),(7.25,1.5,5.5,0.95),
]:
    arrow(ax, xs, ys, xt, yt)

plt.tight_layout()
fig.savefig(FIGS / 'fig1_architecture.png', dpi=150, bbox_inches='tight', facecolor='white')
plt.show()
print('Saved fig1_architecture.png')
```
